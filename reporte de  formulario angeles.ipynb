```python
import pandas as pd
import re
import unicodedata


def normalizar_pregunta(pregunta: str) -> str:
    texto = str(pregunta).strip().lower()
    texto = unicodedata.normalize("NFKD", texto).encode("ascii", "ignore").decode("ascii")
    texto = re.sub(r"_\d+$", "", texto)            # quita sufijo por consultorio (_1, _2, ...)
    texto = re.sub(r"[^a-z0-9]+", "_", texto)       # unifica espacios, comas y simbolos
    texto = re.sub(r"_+", "_", texto).strip("_")   # evita dobles guiones bajos
    return texto


# Informacion de la unidad
unidad = (
    base_an[base_an["tipo_registro"] == "unidad"][
        [
            "entidad",
            "clues_imb",
            "nombre_de_la_unidad",
            "internet",
            "consultorios_habilitados",
        ]
    ]
    .drop_duplicates("clues_imb")
)

# Informacion por consultorio
respuesta = base_an[base_an["tipo_registro"] == "respuesta"].copy()

# Normaliza pregunta para evitar columnas duplicadas por comas/simbolos
respuesta["pregunta"] = respuesta["pregunta"].astype(str)
respuesta["pregunta_norm"] = respuesta["pregunta"].map(normalizar_pregunta)

# Convierte a numerico para consolidar duplicados con max
respuesta["valor_num"] = pd.to_numeric(respuesta["valor"], errors="coerce")

# Convierte preguntas normalizadas en columnas
pivot = respuesta.pivot_table(
    index=["entidad", "clues_imb", "consultorio"],
    columns="pregunta_norm",
    values="valor_num",
    aggfunc="max",
).reset_index()

pivot.columns.name = None

# Unir con la informacion de la unidad
resultado = pivot.merge(
    unidad,
    on=["entidad", "clues_imb"],
    how="left",
)

# Reordenar columnas
columnas_fijas = [
    "entidad",
    "clues_imb",
    "nombre_de_la_unidad",
    "internet",
    "consultorios_habilitados",
    "consultorio",
]

columnas_preguntas = [
    c for c in resultado.columns if c not in columnas_fijas
]

resultado = resultado[columnas_fijas + sorted(columnas_preguntas)]
```

```python
# ============================================
# TABLA 4 - LLENADO POR CLUES Y CONSULTORIO
# Base esperada: preguntas de b (excluyendo 'turno' y filas 0,1,2,5)
# ============================================

import re
import unicodedata


def normalizar_pregunta(x: str) -> str:
    s = str(x).strip().lower()
    s = unicodedata.normalize("NFKD", s).encode("ascii", "ignore").decode("ascii")
    s = re.sub(r"_\d+$", "", s)                # quita sufijo numerico
    s = re.sub(r"[^a-z0-9]+", "_", s)           # espacios/simbolos -> _
    s = re.sub(r"_+", "_", s).strip("_")       # limpia _ repetidos
    s = re.sub(r"_consultorio$", "", s)         # quita sufijo fijo consultorio
    return s

# 1) Catalogo esperado desde b (quitando filas especificas)
b_filtrado = b.drop(index=[0, 1, 2, 5], errors="ignore").copy()

catalogo_b = b_filtrado["preguntas"].dropna().astype(str).str.strip()
catalogo_b = catalogo_b[catalogo_b != ""]
catalogo_b_norm = catalogo_b.map(normalizar_pregunta)
catalogo_b_norm = catalogo_b_norm[catalogo_b_norm.str.contains("turno") == False]
set_catalogo_b = set(catalogo_b_norm.tolist())
n_preguntas_esperadas = len(set_catalogo_b)

print(f"Preguntas esperadas desde b filtrado (sin turno): {n_preguntas_esperadas}")

# 2) Respuestas del formulario por consultorio
resp = base_an[base_an["tipo_registro"] == "respuesta"].copy()
resp["pregunta"] = resp["pregunta"].astype(str).str.strip()
resp["pregunta_norm"] = resp["pregunta"].map(normalizar_pregunta)
resp["consultorio"] = pd.to_numeric(resp["consultorio"], errors="coerce")
resp = resp.dropna(subset=["clues_imb", "consultorio", "pregunta_norm"])
resp = resp[resp["pregunta_norm"] != ""]
resp = resp[resp["pregunta_norm"].str.contains("turno") == False]
resp["consultorio"] = resp["consultorio"].astype(int)

keys = ["entidad", "clues_imb", "nombre_de_la_unidad", "consultorio"]

# 3) Calcula por consultorio: respondidas, faltantes, extras y porcentaje
filas = []
faltantes_detalle = []

for k, grp in resp.groupby(keys, dropna=False):
    preguntas_unicas = set(grp["pregunta_norm"].dropna().astype(str).str.strip())
    preguntas_unicas.discard("")

    respondidas_validas = len(preguntas_unicas & set_catalogo_b)
    faltantes_set = set_catalogo_b - preguntas_unicas
    extras_set = preguntas_unicas - set_catalogo_b

    faltantes = sorted(faltantes_set)
    extras = sorted(extras_set)

    porcentaje = round((respondidas_validas / n_preguntas_esperadas) * 100, 1) if n_preguntas_esperadas else 0.0

    filas.append({
        "entidad": k[0],
        "clues_imb": k[1],
        "nombre_de_la_unidad": k[2],
        "consultorio": k[3],
        "preguntas_esperadas": n_preguntas_esperadas,
        "preguntas_respondidas_validas": respondidas_validas,
        "faltantes": len(faltantes),
        "extras_fuera_catalogo": len(extras),
        "porcentaje_llenado": porcentaje,
        "lista_faltantes": ", ".join(faltantes),
        "lista_extras": ", ".join(extras),
    })

    for pregunta_faltante in faltantes:
        faltantes_detalle.append({
            "entidad": k[0],
            "clues_imb": k[1],
            "nombre_de_la_unidad": k[2],
            "consultorio": k[3],
            "pregunta_faltante": pregunta_faltante,
        })

tabla_llenado_consultorio = pd.DataFrame(filas).sort_values(
    ["clues_imb", "consultorio"],
    ascending=[True, True]
).reset_index(drop=True)

# 4) Detalle de faltantes por consultorio
faltantes_por_consultorio = pd.DataFrame(faltantes_detalle).sort_values(
    ["clues_imb", "consultorio", "pregunta_faltante"],
    ascending=[True, True, True]
).reset_index(drop=True)

# 5) Resumen por CLUES (acumulado de consultorios)
resumen_llenado_clues = (
    tabla_llenado_consultorio
    .groupby(["entidad", "clues_imb", "nombre_de_la_unidad"], as_index=False)
    .agg(
        consultorios_reportados=("consultorio", "nunique"),
        preguntas_esperadas_por_consultorio=("preguntas_esperadas", "max"),
        preguntas_respondidas_validas_total=("preguntas_respondidas_validas", "sum"),
        faltantes_total=("faltantes", "sum"),
        extras_fuera_catalogo_total=("extras_fuera_catalogo", "sum"),
    )
)

resumen_llenado_clues["preguntas_esperadas_total"] = (
    resumen_llenado_clues["consultorios_reportados"]
    * resumen_llenado_clues["preguntas_esperadas_por_consultorio"]
)

resumen_llenado_clues["porcentaje_llenado_total"] = (
    resumen_llenado_clues["preguntas_respondidas_validas_total"]
    / resumen_llenado_clues["preguntas_esperadas_total"]
    * 100
).round(1)

print("\nVista por consultorio (primeras filas):")
display(tabla_llenado_consultorio.head())

print("\nDetalle de faltantes con consultorio (primeras filas):")
display(faltantes_por_consultorio.head())
```

```python
# Consultorios con >= 90% de llenado
tabla_llenado_90 = tabla_llenado_consultorio[
    tabla_llenado_consultorio["porcentaje_llenado"] >= 80
].reset_index(drop=True)

print(f"Consultorios con >= 90% de llenado: {len(tabla_llenado_90)}")

# Faltantes solo de esos consultorios
clues_cons_90 = tabla_llenado_90[["clues_imb", "consultorio"]]

faltantes_90 = faltantes_por_consultorio.merge(
    clues_cons_90,
    on=["clues_imb", "consultorio"],
    how="inner"
).reset_index(drop=True)

print(f"Registros de faltantes en consultorios >= 90%: {len(faltantes_90)}")

display(tabla_llenado_90)
display(faltantes_90)
```

```python
# Agrupar censo_filtrado por entidad, clues_imb y rama con suma de turnos
cols_group = ["entidad", "clues_imb", "rama"]
faltantes = [c for c in cols_group + ["turnos"] if c not in censo_filtrado.columns]
if faltantes:
    raise KeyError(f"Faltan columnas en censo_filtrado: {faltantes}")

tmp = censo_filtrado.copy()
for c in cols_group:
    tmp[c] = tmp[c].astype(str).str.strip()

# Guardar version textual de turnos por fila
tmp["turnos_texto_fila"] = (
    tmp["turnos"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.split(",")
    .apply(lambda xs: ", ".join(sorted({x.strip() for x in xs if x and x.strip()})))
)

# Guardar fuente_de_pago en texto por fila (si existe)
if "fuente_de_pago" in tmp.columns:
    tmp["fuente_de_pago_texto"] = (
        tmp["fuente_de_pago"]
        .fillna("")
        .astype(str)
        .str.strip()
    )
else:
    tmp["fuente_de_pago_texto"] = ""

# 1) Intentar tratar turnos como numerico
turnos_num = pd.to_numeric(tmp["turnos"], errors="coerce")

if turnos_num.notna().any():
    tmp["turnos_num"] = turnos_num.fillna(0)
else:
    # 2) Si turnos viene como texto tipo "lunes-matutino, martes-matutino",
    # contar turnos unicos por fila
    tmp["turnos_num"] = (
        tmp["turnos"]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.split(",")
        .apply(lambda xs: len({x.strip() for x in xs if x and x.strip()}))
    )

# Evitar duplicados por persona-turno antes de sumar
persona_col = next((c for c in ["rfc", "curp", "matricula", "id_personal", "nombre"] if c in tmp.columns), None)
if persona_col:
    dedup_cols = cols_group + [persona_col, "turnos_texto_fila", "fuente_de_pago_texto"]
else:
    dedup_cols = cols_group + ["turnos_texto_fila", "fuente_de_pago_texto"]

tmp_unico = tmp.drop_duplicates(subset=dedup_cols)

censo_agrupado = (
    tmp_unico
    .groupby(cols_group, as_index=False)
    .agg(
        turnos=("turnos_num", "sum"),
        turnos_escritos=(
            "turnos_texto_fila",
            lambda s: ", ".join(
                sorted({
                    tok.strip()
                    for val in s
                    for tok in str(val).split(",")
                    if tok and tok.strip()
                })
            ),
        ),
        fuente_de_pago=(
            "fuente_de_pago_texto",
            lambda s: ", ".join(sorted({v.strip() for v in s if str(v).strip()})),
        ),
        registros=("turnos_num", "size"),
    )
    .sort_values(cols_group)
    .reset_index(drop=True)
)
```

```python
# Consolidar ramas médicas en una sola categoría: MEDICINA
ramas_medicina = [
    "MEDICINA GENERAL",
    "MEDICOS ESPECIALISTAS",
    "MEDICINA TRADICIONAL",
]

# Mantener solo ramas de interés
censo_agrupado = censo_agrupado[
    censo_agrupado["rama"].isin(["ENFERMERIA", *ramas_medicina])
] .copy()

# Filas no médicas (por ahora solo ENFERMERIA)
no_medicina = censo_agrupado[~censo_agrupado["rama"].isin(ramas_medicina)].copy()

# Filas médicas consolidadas
medicina = (
    censo_agrupado[censo_agrupado["rama"].isin(ramas_medicina)]
    .assign(rama="MEDICINA")
    .groupby(["entidad", "clues_imb", "rama"], as_index=False)
    .agg(
        turnos=("turnos", "sum"),
        turnos_escritos=(
            "turnos_escritos",
            lambda s: ", ".join(
                sorted({
                    tok.strip()
                    for val in s
                    for tok in str(val).split(",")
                    if tok and tok.strip()
                })
            ),
        ),
        fuente_de_pago=(
            "fuente_de_pago",
            lambda s: ", ".join(sorted({v.strip() for v in s if str(v).strip()})),
        ),
        registros=("registros", "sum"),
    )
)

# Resultado final
censo_agrupado = (
    pd.concat([no_medicina, medicina], ignore_index=True)
    .sort_values(["entidad", "clues_imb", "rama"])
.reset_index(drop=True)
)
```

```python
# Pivotear rama para tener columnas por rama y sus datos
censo_pivot = censo_agrupado.pivot_table(
    index=["entidad", "clues_imb"],
    columns="rama",
    values=["turnos", "turnos_escritos", "fuente_de_pago", "registros"],
    aggfunc="first",
)

# Aplanar nombres de columnas
censo_pivot.columns = [f"{metrica}_{rama}" for metrica, rama in censo_pivot.columns]
censo_pivot = censo_pivot.reset_index()

# Rellenar nulos segun tipo de columna
for col in censo_pivot.columns:
    if col.startswith(("turnos_escritos_", "fuente_de_pago_")):
        censo_pivot[col] = censo_pivot[col].fillna("")
    elif col.startswith(("turnos_", "registros_")):
        censo_pivot[col] = pd.to_numeric(censo_pivot[col], errors="coerce").fillna(0)

# Homologacion de turnos: quitar dia y conservar solo turno
def homologar_turnos(texto):
    tokens = [t.strip().lower() for t in str(texto).split(",") if t and t.strip()]

    categorias = set()
    for t in tokens:
        if "matutino" in t:
            categorias.add("Matutino")
        elif "vespertino" in t:
            categorias.add("Vespertino")
        elif "nocturno" in t:
            categorias.add("Nocturno")
        elif "jornadaacumulada" in t or "jornada acumulada" in t:
            categorias.add("Jornada acumulada")

    salida = []
    if "Matutino" in categorias and "Vespertino" in categorias:
        salida.append("Ambos")
        categorias.discard("Matutino")
        categorias.discard("Vespertino")
    else:
        if "Matutino" in categorias:
            salida.append("Matutino")
        if "Vespertino" in categorias:
            salida.append("Vespertino")

    if "Nocturno" in categorias:
        salida.append("Nocturno")
    # Jornada acumulada se conserva en el dato original, pero no se muestra en homologados

    return ", ".join(salida)

for col in [c for c in censo_pivot.columns if c.startswith("turnos_escritos_")]:
    rama = col.replace("turnos_escritos_", "")
    censo_pivot[f"turnos_homologados_{rama}"] = censo_pivot[col].apply(homologar_turnos)

censo_pivot.head()
```
